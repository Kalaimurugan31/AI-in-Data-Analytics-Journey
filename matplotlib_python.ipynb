```python
import pandas as pd
import matplotlib.pyplot as plt
data={
    "name":["kalai","kesavan","kayal","vicky","naveen","akbar"],
    "age":[20,24,23,25,24,23]

}
df=pd.DataFrame(data)
plt.bar(df["name"], df["age"])
plt.title("age")
plt.xlabel("age")
plt.ylabel("name")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data={
    "dept": ["IT","Sales","HR","Finance","Marketing"],
    "employees": [30,25,15,20,10]
}
df=pd.DataFrame(data)
plt.pie(df["employees"],
   labels=df["dept"],
        autopct="%1.1f%%",
        colors=["yellow","lightyellow","blue","orange","red"])
plt.title("employees per department")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data={
    "name":["kalai","kesavan","kayal","vicky","saravanan","nithish","naveen"],
    "salary":[25000,30000,20000,15000,22000,26000,24000]

}
df=pd.DataFrame(data)
plt.hist(df["salary"], bins=10,color="blue")
plt.title("salary distribution")
plt.xlabel("salary range")
plt.ylabel("number of employees")
plt.show()
```

```python
##scatter chart
import pandas as pd
import matplotlib.pyplot as plt
data={
    "name":["kalai","kesavan","kayal","vicky","saravanan","nithish","naveen"],
    "age":[20,25,26,24,26,25,23],
    "salary":[25000,15000,26000,30000,28000,27000,28000]

}
df=pd.DataFrame(data)
plt.scatter(df["age"],df["salary"])
plt.title("salary and their age")
plt.xlabel("age")
plt.ylabel("name")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data={
    "name":["kalai","naveen","chandru","karhika","akshitha"],
    "age":[20,26,24,23,27],
    "salary":[25000,24000,23000,18000,24000]
}
df=pd.DataFrame(data)
fig,(ax1,ax2)=plt.subplots(1,2,figsize=(12,10))
ax1.bar(df["name"],df["salary"],color="red")
ax1.set_title("employee comparision")
ax1.set_xlabel("name")
ax1.set_ylabel("salary")

##chart2
ax2.scatter(df["age"],df["salary"],color="blue")
ax2.set_title("age vs salary")
ax2.set_xlabel("age")
ax2.set_ylabel("salary")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data={
    "customers":[100,101,102,103,104,105,106,107,108,109,110],
    "sales":[10000,15000,12000,14000,16000,18000,14000,16000,15000,14000,15000],
    "expenses":[5000,4000,3000,6000,2000,4000,2500,6000,1400,5000,4000],
    "profit":[5000,11000,9000,8000,14000,14000,11500,10000,13600,9000,11000],
    "month":["jan","feb","mar","apr","may","jun","jul","aug","sep","oct","nov"]
}
df=pd.DataFrame(data)
fig,((ax1,ax2),(ax3,ax4))=plt.subplots(2,2,figsize=(12,5))
##barchart
ax1.bar(df["month"],df["profit"])
ax1.set_title("profit per month")
ax1.set_xlabel("month")
ax1.set_ylabel("profit")
##line(plot)
ax2.plot(df["month"],df["sales"],color="red",marker="o",label="sales")
ax2.set_title("month wise sale")
ax2.set_xlabel("month")
ax2.set_ylabel("sales")
ax2.legend()
##piechart
ax3.pie(df["expenses"],labels=df["month"],autopct="%1.1f%%")
ax3.set_title("expenses")

##scatter
ax4.scatter(df["customers"],df["profit"])
ax4.set_title("customers and their profit")
ax4.set_xlabel("customers")
ax4.set_ylabel("profit")

plt.tight_layout()
plt.show()
```
