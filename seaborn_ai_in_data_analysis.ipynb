```python
##seaborn
##barplot
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data={
    "name": ["kalai","naveen","chandru","karhika","akshitha",
             "naresh","karthi","sathish"],
    "dept": ["it","sales","hr","it","sales","hr","it","sales"],
    "salary": [26000,25000,28000,24000,36000,26000,15000,20000],
    "age": [26,25,24,29,26,25,26,24]
}
df=pd.DataFrame(data)
sns.barplot(x="name",y="salary",data=df)
plt.title("employee salary")
plt.xlabel("name")
plt.ylabel("salary")
plt.show()
```

```python
##scatter plot
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "name": ["kalai","naveen","chandru","karhika","akshitha",
             "naresh","karthi","sathish"],
    "dept": ["it","sales","hr","it","sales","hr","it","sales"],
    "salary": [26000,25000,28000,24000,36000,26000,15000,20000],
    "age": [26,25,24,29,26,25,26,24]
}
df = pd.DataFrame(data)

sns.scatterplot(x="age", y="salary", hue="dept", data=df)
plt.title("age vs salary by dept")
plt.xlabel("age")
plt.ylabel("Salary")
plt.show()
```

```python
##heatmap
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data={
     "name": ["kalai","naveen","chandru","karhika","akshitha",
             "naresh","karthi","sathish"],
    "dept": ["it","sales","hr","it","sales","hr","it","sales"],
    "salary": [26000,25000,28000,24000,36000,26000,15000,20000],
    "age": [26,25,24,29,26,25,26,24]

}
df=pd.DataFrame(data)
correlation=df[["age","salary"]].corr()
sns.heatmap(correlation,annot=True,cmap="coolwarm")
plt.title("correlation heatmap")
plt.show()

```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data={
    "name": ["kalai","naveen","chandru","karhika","akshitha",
             "naresh","karthi","sathish","raman","pradeep"],
    "age": [26,25,24,29,26,25,26,24,23,27],
    "dept": ["it","sales","hr","it","sales","hr","it","sales","hr","it"],
    "salary": [26000,25000,28000,24000,36000,26000,15000,20000,22000,30000],
    "experience": [2,3,1,5,4,2,1,3,2,4]
}

df=pd.DataFrame(data)
fig,((ax1,ax2),(ax3,ax4))=plt.subplots(2,2, figsize=(10,12))
##chart 1
sns.barplot(x="name",y="salary",data=df,ax=ax1)
ax1.set_title("employee and their salary")
ax1.set_xlabel("employee")
ax1.set_ylabel("salary")
##chart 2
sns.boxplot(x="dept",y="salary",data=df,ax=ax2)
ax2.set_title("dept wise salary")
ax2.set_xlabel("dept")
ax2.set_ylabel("salary")
##chart3
sns.scatterplot(x="age",y="salary",hue="dept",data=df,ax=ax3)
ax3.set_title("does age affect salaey")
ax3.set_xlabel("age")
ax3.set_ylabel("salary")
##chart 4
correlation=df[["age","salary","experience"]].corr()
sns.heatmap(correlation,annot=True,cmap="coolwarm",ax=ax4)
ax4.set_title("correlation heatmap")
plt.tight_layout()
plt.show()

```
