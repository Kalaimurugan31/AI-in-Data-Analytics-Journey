```python
##statistics-class 1 mean median mode
import numpy as np
marks=np.array([45,50,60,72,96,78,96,40,20,])
mean=np.mean(marks)
print("mean:",mean)
##median
median=np.median(marks)
print("median:",median)


from scipy import stats
mode=stats.mode(marks)
print("mode:",mode)

import numpy as np
from scipy import stats
salary=np.array([20000,15000,30000,25000,36000,25000,15000,20000,30000,24000])
print("mean:",np.mean(salary))
print("median:",np.median(salary))
print("mode:",stats.mode(salary))
```

```python
##class 2 variance and standard deviation
import numpy as np
salary = np.array([20000,15000,30000,25000,36000,
                   25000,15000,20000,30000,24000])
print("mean:",np.mean(salary))
print("variance:",np.var(salary))
print("std:",np.std(salary))

mean=np.mean(salary)
std=np.std(salary)

print("most salaries fall between:")
print("lower:",mean-std)
print("upper:",mean+std)
```

```python
import numpy as np
score=np.array([76,86,50,78,96,45,63,46,70,100])
##calculating the total
total=len(score)
print("total score:",total)
##probability of getting score above >70
above_70=len(score[score>70])
prob_above_70=above_70/total
print("probablitiy of getting above 70:",prob_above_70)

##calculating total
total=len(score)
print("total score:",total)
##probablity of getting score below <60
below_60=len(score[score<60])
prob_below_60=below_60/total
print("probablity of getting below 60:",prob_below_60)
##total score
total=len(score)
print("total score:",total)
##probablity of getting = 78
equal_78=len(score[score==78])
prob_equal_78=equal_78/total
print("probablity of getting =78:",prob_equal_78)
```

```python
import numpy as np
import matplotlib.pyplot as plt
##normal distribution data
data=np.random.normal(loc=24000,scale=5000,size=1000)
##ploting
plt.hist(data,bins=30,color="blue",edgecolor="black")
plt.title("employee salary distribution")
plt.xlabel("salary")
plt.ylabel("number of employees")
plt.show()
```

```python
  ##statistics
print("mean:",np.mean(data))
print("std:",np.std(data))
print("min:",np.min(data))
print("max:",np.max(data))

  ##68-95-99 rule
mean=np.mean(data)
std=np.std(data)
print("68% of data falls between:",mean-std,"and",mean+std)
print("95% of data falls between:",mean-(2*std),"and",mean+(2*std))
print("99% of data falls between:",mean-(3*std),"and",mean+(3*std))

```

```python
##Statistics — Class 5: Correlation & Regression
## What is Correlation?
##Measures how strongly two things are related!Value between -1 and 1:

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data={
    "age":[22,25,26,28,24,23,27,29,25,26],
    "experience":[1,3,4,6,2,1,5,7,3,4],
    "salary":[20000,25000,28000,35000,22000,
               20000,30000,38000,25000,28000],
    "shoe size":[5,3,2,1,4,6,2,1,3,2]
}

df=pd.DataFrame(data)
##correlation matrix
correlation=df.corr()
print(correlation)
##heat map chart
sns.heatmap(correlation,annot=True,cmap="coolwarm")
plt.title("correlation heatmap")
plt.show()

import numpy as np
import matplotlib.pyplot as plt


age=np.array([22,25,26,28,24,23,27,29,25,26])
salary=np.array([20000,25000,28000,35000,22000,
                   20000,30000,38000,25000,28000])
##array for particularly two columns
correlation=np.corrcoef(age,salary)
print("correlation vs age:")
print(correlation)
plt.scatter(age,salary)
plt.plot(np.unique(age),np.poly1d(np.polyfit(age,salary,1))(np.unique(age)),color="red")
plt.show()
```
