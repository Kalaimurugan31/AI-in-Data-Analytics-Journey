```python
import numpy as np
arr=np.array([10,20,30,40,50])
##add five to every element
print(arr+5)
##multiply every element
print(arr*2)
##divide every element
print(arr/2)
##square every element
print(arr**2)
```

```python
##math between two arrays
arr1=np.array([10,20,30,40,50])
arr2=np.array([1,2,3,4,5])
##add two arrays
print(arr1+arr2)
##multiply two arrays
print(arr1*arr2)
##subtarct two arrays
print(arr1-arr2)
```

```python
##built math function
arr5=np.array([10,20,30,40,50])
print(np.sum(arr5))
print(np.mean(arr5))
print(np.min(arr5))
print(np.max(arr5))
print(np.std(arr5))
```

```python
##class5-numpy with pandas
import pandas as pd
import numpy as np
data={
    "name":["kalai","kesavan","vicky","kayal","saravana","nithish","anu","naveen"],
    "age":[25,30,20,23,24,23,23,24],
    "salary":[25000,30000,20000,15000,24000,23000,23000,24000]
}
df=pd.DataFrame(data)
##convert salary column to numpy
salary_arr=np.array(df["salary"])
print(salary_arr)

##calculate using numpy
print("sum:",np.sum(salary_arr))
print("mean:",np.mean(salary_arr))
print("max:",np.max(salary_arr))
print("min:",np.min(salary_arr))
print("std:",np.std(salary_arr))

##bonus
bonus=np.array([1000,5000,2000,3000,6000,4500,4000,2500])
df["total_salary"]=salary_arr+bonus
print(df)
```
